```python
# =========================
# 📊 Netflix Dataset Analysis in Google Colab
# =========================

# Step 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Mount Google Drive (optional if file is in Drive)
# from google.colab import drive
# drive.mount('/content/drive')

# Step 3: Upload the file manually or use path if already uploaded
# If you uploaded manually in Colab:
from google.colab import files
uploaded = files.upload()

# Step 4: Load Dataset
# Update the file path below if you are not using the manual upload or if the file name is different
file_name = list(uploaded.keys())[0] if uploaded else "/content/Netflix Dataset (1).csv"
df = pd.read_csv(file_name)

# Step 5: Display first few rows
print("🔹 First 5 rows of the dataset:")
display(df.head())

# Step 6: Basic Information
print("\n🔹 Dataset Info:")
df.info()

print("\n🔹 Missing Values:")
print(df.isnull().sum())

print("\n🔹 Summary Statistics:")
display(df.describe(include='all'))

# Step 7: Basic Analysis
print("\n🔹 Number of Movies vs TV Shows:")
print(df['Category'].value_counts())

# Plot Movies vs TV Shows
df['Category'].value_counts().plot(kind='bar', color=['red', 'blue'])
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()

# Step 8: Most Common Genres (if column exists)
if 'listed_in' in df.columns:
    genres = df['listed_in'].dropna().apply(lambda x: x.split(',')).explode().str.strip()
    top_genres = genres.value_counts().head(10)
    print("\n🔹 Top 10 Genres:")
    display(top_genres)

    top_genres.plot(kind='barh', figsize=(8,5))
    plt.title("Top 10 Genres on Netflix")
    plt.xlabel("Count")
    plt.ylabel("Genre")
    plt.gca().invert_yaxis()
    plt.show()

# Step 9: Content Added per Year (if release_year/date_added exists)
if 'release_year' in df.columns:
    df['release_year'].value_counts().sort_index().plot(kind='line', figsize=(10,5))
    plt.title("Content Added per Year")
    plt.xlabel("Year")
    plt.ylabel("Number of Titles")
    plt.show()

print("\n✅ Analysis Completed Successfully!")
```

## Analyze content distribution by rating

### Subtask:
Determine the distribution of content ratings in the dataset.

**Reasoning**:
Count the occurrences of each rating and store the result in `rating_counts`. Then, print and visualize the result.

```python
# Analyze content distribution by rating
rating_counts = df['Rating'].value_counts()

print("\n🔹 Content Distribution by Rating:")
display(rating_counts)

# Plot the rating distribution
plt.figure(figsize=(10, 6))
rating_counts.plot(kind='bar')
plt.title("Content Distribution by Rating on Netflix")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Step 10: Content by Country (if column exists)
if 'country' in df.columns:
    # Some entries have multiple countries, so we need to split and count
    countries = df['country'].dropna().apply(lambda x: x.split(',')).explode().str.strip()

    # Get the top 10 countries with the most content
    top_countries = countries.value_counts().head(10)

    print("\n🔹 Top 10 Countries by Content:")
    display(top_countries)

    # Plot the top 10 countries
    top_countries.plot(kind='barh', figsize=(8, 5))
    plt.title("Top 10 Countries by Content on Netflix")
    plt.xlabel("Number of Titles")
    plt.ylabel("Country")
    plt.gca().invert_yaxis() # Invert y-axis to show the highest count at the top
    plt.show()
else:
    print("\n🔹 'country' column not found in the dataset.")
```

# Task
Analyze the relationship between directors and content in the dataset.

## Identify top directors

### Subtask:
Determine the directors with the most titles in the dataset.


**Reasoning**:
Count the occurrences of each director, get the top 10, and store the result in `top_directors`. Then, print the result.



```python
top_directors = df['Director'].value_counts().head(10)
print("\n🔹 Top 10 Directors by Content:")
display(top_directors)
```

## Analyze content type by director

### Subtask:
For the top directors, analyze the distribution of content types (movies vs. TV shows) they have directed.


**Reasoning**:
Filter the dataframe to include only titles directed by the top directors and then group by director and category to count the number of movies and TV shows for each top director.



```python
# Filter the DataFrame to include only titles directed by the top directors
top_directors_list = top_directors.index.tolist()
filtered_df = df[df['Director'].isin(top_directors_list)].copy()

# Group by director and category, then count the number of titles
director_category_counts = filtered_df.groupby(['Director', 'Category']).size().unstack(fill_value=0)

print("\n🔹 Content Type Distribution for Top Directors:")
display(director_category_counts)
```

## Visualize findings

### Subtask:
Create visualizations to show the top directors and the types of content they direct.


**Reasoning**:
Create a stacked bar chart to visualize the content type distribution for the top directors using the `director_category_counts` DataFrame.



```python
director_category_counts.plot(kind='bar', stacked=True, figsize=(12, 7))
plt.title("Content Type Distribution for Top Directors")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for readability
plt.tight_layout() # Adjust layout to prevent labels overlapping
plt.show()
```
